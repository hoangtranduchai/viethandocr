```python
import os
import json
import glob
import random
import pandas as pd

# Cố định random seed để kết quả phân chia có tính lặp lại (reproducible)
random.seed(42)

output_dir = "/kaggle/working/vietocr_data"
os.makedirs(output_dir, exist_ok=True)

# -------------------------------------------------------------
# 1. TRÍCH XUẤT DỮ LIỆU TỪ UIT-HWDB-LINE
# -------------------------------------------------------------
uit_base = "/kaggle/input/datasets/danhngnnguyn/uit-hwdb-line/UIT_HWDB_line"
uit_writers = glob.glob(f"{uit_base}/*/*/")
uit_writers = [w for w in uit_writers if os.path.exists(os.path.join(w, "label.json"))]

# Dictionary lưu: {writer_unique_id: [line_string, line_string, ...]}
dataset_by_writer = {}

print(f"-> Đang xử lý UIT-HWDB: phát hiện {len(uit_writers)} tác giả...")
for w_dir in uit_writers:
    # Định danh duy nhất cho writer của UIT
    writer_key = f"uit_{os.path.basename(os.path.normpath(w_dir))}"
    dataset_by_writer[writer_key] = []
    
    json_path = os.path.join(w_dir, "label.json")
    try:
        with open(json_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            for img_name, text in data.items():
                clean_text = " ".join(str(text).strip().split())
                if not clean_text:
                    continue
                img_path = os.path.join(w_dir, img_name)
                if os.path.exists(img_path):
                    dataset_by_writer[writer_key].append(f"{img_path}\t{clean_text}")
    except Exception as e:
        print(f"Lỗi đọc {json_path}: {e}")

# -------------------------------------------------------------
# 2. TRÍCH XUẤT DỮ LIỆU TỪ KALAPA (Training annotations)
# -------------------------------------------------------------
kalapa_base = "/kaggle/input/datasets/danhngnnguyn/kalapa-ocr/OCR"
kalapa_train_img_root = os.path.join(kalapa_base, "training_data/training_data/images")
kalapa_anno_dir = os.path.join(kalapa_base, "training_data/training_data/annotations")

kalapa_annos = glob.glob(f"{kalapa_anno_dir}/*.txt")
print(f"-> Đang xử lý KALAPA Training: phát hiện {len(kalapa_annos)} file nhãn...")

for anno_file in kalapa_annos:
    folder_id = os.path.splitext(os.path.basename(anno_file))[0]
    writer_key = f"kalapa_{folder_id}"
    if writer_key not in dataset_by_writer:
        dataset_by_writer[writer_key] = []
        
    with open(anno_file, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split('\t')
            if len(parts) >= 2:
                rel_img_path = parts[0].strip()
                text = " ".join(parts[1].strip().split())
                full_img_path = os.path.join(kalapa_train_img_root, rel_img_path)
                if os.path.exists(full_img_path) and text:
                    dataset_by_writer[writer_key].append(f"{full_img_path}\t{text}")

# -------------------------------------------------------------
# 3. TRÍCH XUẤT DỮ LIỆU TỪ KALAPA (Test label CSV nếu có ảnh)
# -------------------------------------------------------------
test_csv_path = os.path.join(kalapa_base, "ocr_private_test_label_cleaned.csv")
# Thư mục ảnh test nếu có trong dataset (thường là Level2_samples hoặc test_images)
test_img_dir = os.path.join(kalapa_base, "Level2_samples")

if os.path.exists(test_csv_path) and os.path.exists(test_img_dir):
    print("-> Đang kiểm tra và thêm nhãn từ KALAPA Test CSV...")
    df_test = pd.read_csv(test_csv_path)
    for _, row in df_test.iterrows():
        rel_path = str(row['id']).strip()
        text = " ".join(str(row['answer']).strip().split())
        full_path = os.path.join(test_img_dir, rel_path)
        
        # Nếu tồn tại ảnh thực tế thì thêm vào
        if os.path.exists(full_path) and text and text.lower() != 'nan':
            folder_id = rel_path.split('/')[0] if '/' in rel_path else "root"
            writer_key = f"kalapa_test_{folder_id}"
            if writer_key not in dataset_by_writer:
                dataset_by_writer[writer_key] = []
            dataset_by_writer[writer_key].append(f"{full_path}\t{text}")

# -------------------------------------------------------------
# 4. CHIA TRAIN / VAL THEO TÁC GIẢ (WRITER-INDEPENDENT 90/10)
# -------------------------------------------------------------
# Loại bỏ những writer không có dòng ảnh nào hợp lệ
all_writers = [w for w, lines in dataset_by_writer.items() if len(lines) > 0]
random.shuffle(all_writers)

split_idx = int(len(all_writers) * 0.9)
train_writers = all_writers[:split_idx]
val_writers = all_writers[split_idx:]

train_lines = []
for w in train_writers:
    train_lines.extend(dataset_by_writer[w])

val_lines = []
for w in val_writers:
    val_lines.extend(dataset_by_writer[w])

# Trộn ngẫu nhiên thứ tự các dòng trong từng tập để tối ưu cho DataLoader
random.shuffle(train_lines)
random.shuffle(val_lines)

# -------------------------------------------------------------
# 5. XUẤT FILE CHO VIETOCR
# -------------------------------------------------------------
train_label_file = os.path.join(output_dir, "train_label.txt")
val_label_file = os.path.join(output_dir, "val_label.txt")

with open(train_label_file, 'w', encoding='utf-8') as f:
    f.write("\n".join(train_lines) + "\n")

with open(val_label_file, 'w', encoding='utf-8') as f:
    f.write("\n".join(val_lines) + "\n")

print("\n================= TỔNG KẾT DỮ LIỆU ETL =================")
print(f"Tổng số nhóm tác giả (Writers): {len(all_writers)}")
print(f"Số nhóm tác giả tập Train: {len(train_writers)} | Validation: {len(val_writers)}")
print(f"Tổng số mẫu Train (train_label.txt): {len(train_lines):,} dòng")
print(f"Tổng số mẫu Validation (val_label.txt): {len(val_lines):,} dòng")
print(f"Tỷ lệ mẫu thực tế: {len(train_lines)/(len(train_lines)+len(val_lines))*100:.2f}% Train / {len(val_lines)/(len(train_lines)+len(val_lines))*100:.2f}% Val")
```

```python
import os

train_file = "/kaggle/working/vietocr_data/train_label.txt"
val_file = "/kaggle/working/vietocr_data/val_label.txt"

def extract_metadata(file_path):
    writers = set()
    image_paths = set()
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            img_path = line.split('\t')[0]
            image_paths.add(img_path)
            
            # Trích xuất Folder ID / Writer ID từ đường dẫn
            # Cấu trúc path: .../<writer_id>/<image_name>.jpg
            parts = os.path.normpath(img_path).split(os.sep)
            writer_id = parts[-2]
            
            # Phân biệt rõ writer của uit hay kalapa dựa trên đường dẫn
            dataset_prefix = "uit" if "uit-hwdb" in img_path.lower() else "kalapa"
            writers.add(f"{dataset_prefix}_{writer_id}")
            
    return writers, image_paths

train_writers, train_imgs = extract_metadata(train_file)
val_writers, val_imgs = extract_metadata(val_file)

# 1. KIỂM CHỨNG TRÙNG LẶP TÁC GIẢ (WRITER OVERLAP)
overlap_writers = train_writers.intersection(val_writers)

# 2. KIỂM CHỨNG TRÙNG LẶP ĐƯỜNG DẪN ẢNH (IMAGE OVERLAP)
overlap_images = train_imgs.intersection(val_imgs)

print("============ BÁO CÁO KIỂM CHỨNG ĐỘ TOÀN VẸN DỮ LIỆU ============")
print(f"Tổng số tác giả tập Train: {len(train_writers)}")
print(f"Tổng số tác giả tập Val:   {len(val_writers)}")
print(f"-> Số tác giả bị trùng lặp giữa 2 tập: {len(overlap_writers)}")

print("-" * 65)
print(f"Tổng số ảnh tập Train: {len(train_imgs)}")
print(f"Tổng số ảnh tập Val:   {len(val_imgs)}")
print(f"-> Số ảnh bị trùng lặp giữa 2 tập:     {len(overlap_images)}")
print("=================================================================")

if len(overlap_writers) == 0 and len(overlap_images) == 0:
    print("✅ XÁC NHẬN: Phân chia chuẩn Writer-Independent 100%! Không có rò rỉ dữ liệu (Zero Data Leakage).")
else:
    print("❌ CẢNH BÁO: Phát hiện trùng lặp!")
    if overlap_writers:
        print(f"Các tác giả bị trùng: {overlap_writers}")
```

```python
import os

train_file = "/kaggle/working/vietocr_data/train_label.txt"
val_file = "/kaggle/working/vietocr_data/val_label.txt"

def get_writers_and_images(file_path):
    writers = set()
    images = set()
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            img_path = line.split('\t')[0]
            images.add(img_path)
            
            # Chuẩn hóa đường dẫn để bóc tách
            norm_p = os.path.normpath(img_path)
            parts = norm_p.split(os.sep)
            
            # Lấy tên folder người viết và định danh rõ nguồn
            folder_id = parts[-2]
            if "uit-hwdb" in norm_p.lower():
                # Phân biệt rõ train/test của uit
                split_tag = "train" if "train_data" in norm_p else "test"
                writers.add(f"uit_{split_tag}_{folder_id}")
            elif "kalapa" in norm_p.lower():
                writers.add(f"kalapa_{folder_id}")
            else:
                writers.add(f"unknown_{folder_id}")
                
    return writers, images

train_writers, train_imgs = get_writers_and_images(train_file)
val_writers, val_imgs = get_writers_and_images(val_file)

overlap_writers = train_writers.intersection(val_writers)
overlap_imgs = train_imgs.intersection(val_imgs)

print("=== KẾT QUẢ KIỂM TRA HIỆN TRẠNG ===")
print(f"Tổng số writer tập Train: {len(train_writers)}")
print(f"Tổng số writer tập Val:   {len(val_writers)}")
print(f"Trùng lặp Writer giữa 2 tập: {len(overlap_writers)}")
print(f"Trùng lặp Ảnh giữa 2 tập:    {len(overlap_imgs)}")

if len(overlap_writers) > 0:
    print(f"⚠️ CẢNH BÁO: Có {len(overlap_writers)} writer bị trùng lặp!")
    print(list(overlap_writers)[:5])
else:
    print("✅ XÁC NHẬN: Không trùng tác giả!")
```

```python
import os
import json
import glob
import random

random.seed(42)

output_dir = "/kaggle/working/vietocr_data"
os.makedirs(output_dir, exist_ok=True)

# -------------------------------------------------------------
# 1. GOM NHÓM DỮ LIỆU UIT-HWDB THEO TỪNG TÁC GIẢ (WRITER)
# -------------------------------------------------------------
uit_base = "/kaggle/input/datasets/danhngnnguyn/uit-hwdb-line/UIT_HWDB_line"
uit_folders = glob.glob(f"{uit_base}/*/*/")
uit_data_by_writer = {}

for w_dir in uit_folders:
    json_path = os.path.join(w_dir, "label.json")
    if not os.path.exists(json_path):
        continue
    
    # Định danh duy nhất: uit_train_7 hoặc uit_test_250
    norm_dir = os.path.normpath(w_dir)
    parent_type = os.path.basename(os.path.dirname(norm_dir))  # train_data hoặc test_data
    w_id = os.path.basename(norm_dir)
    writer_key = f"uit_{parent_type}_{w_id}"
    
    samples = []
    try:
        with open(json_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            for img_name, text in data.items():
                clean_text = " ".join(str(text).strip().split())
                img_path = os.path.join(w_dir, img_name)
                if clean_text and os.path.exists(img_path):
                    samples.append(f"{img_path}\t{clean_text}")
    except Exception as e:
        print(f"Lỗi đọc {json_path}: {e}")
        
    if samples:
        uit_data_by_writer[writer_key] = samples

# -------------------------------------------------------------
# 2. GOM NHÓM DỮ LIỆU KALAPA THEO TỪNG TÁC GIẢ (WRITER / FOLDER ID)
# -------------------------------------------------------------
kalapa_base = "/kaggle/input/datasets/danhngnnguyn/kalapa-ocr/OCR"
kalapa_img_root = os.path.join(kalapa_base, "training_data/training_data/images")
kalapa_anno_dir = os.path.join(kalapa_base, "training_data/training_data/annotations")

kalapa_annos = glob.glob(f"{kalapa_anno_dir}/*.txt")
kalapa_data_by_writer = {}

for anno_file in kalapa_annos:
    w_id = os.path.splitext(os.path.basename(anno_file))[0]
    writer_key = f"kalapa_train_{w_id}"
    samples = []
    
    with open(anno_file, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split('\t')
            if len(parts) >= 2:
                rel_img_path = parts[0].strip()
                text = " ".join(parts[1].strip().split())
                full_img_path = os.path.join(kalapa_img_root, rel_img_path)
                if text and os.path.exists(full_img_path):
                    samples.append(f"{full_img_path}\t{text}")
                    
    if samples:
        kalapa_data_by_writer[writer_key] = samples

# -------------------------------------------------------------
# 3. CHIA WRITER-INDEPENDENT THEO NGUYÊN TẮC STRATIFIED (90/10)
# -------------------------------------------------------------
def split_writers_90_10(writer_dict):
    writers = list(writer_dict.keys())
    random.shuffle(writers)
    idx = int(len(writers) * 0.9)
    train_w = set(writers[:idx])
    val_w = set(writers[idx:])
    return train_w, val_w

uit_train_w, uit_val_w = split_writers_90_10(uit_data_by_writer)
kalapa_train_w, kalapa_val_w = split_writers_90_10(kalapa_data_by_writer)

# Gom danh sách tác giả chung
final_train_writers = uit_train_w.union(kalapa_train_w)
final_val_writers = uit_val_w.union(kalapa_val_w)

# Gom tất cả các dòng ảnh
train_lines = []
for w in uit_train_w:
    train_lines.extend(uit_data_by_writer[w])
for w in kalapa_train_w:
    train_lines.extend(kalapa_data_by_writer[w])

val_lines = []
for w in uit_val_w:
    val_lines.extend(uit_data_by_writer[w])
for w in kalapa_val_w:
    val_lines.extend(kalapa_data_by_writer[w])

# Trộn thứ tự dòng trước khi lưu để model học đa dạng
random.shuffle(train_lines)
random.shuffle(val_lines)

# -------------------------------------------------------------
# 4. XUẤT FILE VÀ KIỂM TOÁN TỨC THÌ
# -------------------------------------------------------------
train_path = os.path.join(output_dir, "train_label.txt")
val_path = os.path.join(output_dir, "val_label.txt")

with open(train_path, 'w', encoding='utf-8') as f:
    f.write("\n".join(train_lines) + "\n")

with open(val_path, 'w', encoding='utf-8') as f:
    f.write("\n".join(val_lines) + "\n")

# KIỂM TOÁN TẬP HỢP TOÁN HỌC NGAY TẠI CHỖ
intersection = final_train_writers.intersection(final_val_writers)

print("\n" + "="*50)
print("=== BÁO CÁO PHÂN CHIA WRITER-INDEPENDENT ===")
print("="*50)
print(f"UIT-HWDB: {len(uit_train_w)} tác giả train | {len(uit_val_w)} tác giả val")
print(f"KALAPA:   {len(kalapa_train_w)} tác giả train | {len(kalapa_val_w)} tác giả val")
print(f"TỔNG CỘNG: {len(final_train_writers)} tác giả train | {len(final_val_writers)} tác giả val")
print(f"Số dòng train_label.txt: {len(train_lines):,} mẫu")
print(f"Số dòng val_label.txt:   {len(val_lines):,} mẫu")
print(f"Giao giữa tập Train và Val: {len(intersection)} tác giả")

if len(intersection) == 0:
    print("🎯 CHỨNG MINH HOÀN TẤT: Phân chia đạt chuẩn Writer-Independent tuyệt đối 100%!")
else:
    print("❌ LỖI: Vẫn còn trùng lặp!")
```

```python
import os

package_dir = "/kaggle/working/vietocr_data_package"
images_dir = os.path.join(package_dir, "images")
labels_dir = os.path.join(package_dir, "labels")

os.makedirs(images_dir, exist_ok=True)
os.makedirs(labels_dir, exist_ok=True)

# 1. Tạo symlink liên kết thư mục ảnh gốc vào package
# (Symlink giúp truy cập ảnh như thư mục nội bộ mà không tốn dung lượng đĩa)
uit_src = "/kaggle/input/datasets/danhngnnguyn/uit-hwdb-line/UIT_HWDB_line"
kalapa_src = "/kaggle/input/datasets/danhngnnguyn/kalapa-ocr/OCR/training_data/training_data/images"

uit_link = os.path.join(images_dir, "uit_hwdb")
kalapa_link = os.path.join(images_dir, "kalapa")

if not os.path.exists(uit_link):
    os.symlink(uit_src, uit_link)
if not os.path.exists(kalapa_link):
    os.symlink(kalapa_src, kalapa_link)

print("Đã tạo liên kết dữ liệu ảnh thành công!")

# 2. Chuẩn hóa lại đường dẫn trong train_label.txt và val_label.txt thành đường dẫn tương đối
def convert_to_relative(input_txt, output_txt):
    converted_lines = []
    with open(input_txt, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split('\t')
            abs_path, label = parts[0], parts[1]
            
            # Chuyển đổi đường dẫn tuyệt đối sang dạng: images/uit_hwdb/... hoặc images/kalapa/...
            if "uit-hwdb" in abs_path.lower():
                rel_suffix = abs_path.split("UIT_HWDB_line/")[-1]
                clean_path = f"images/uit_hwdb/{rel_suffix}"
            elif "kalapa" in abs_path.lower():
                rel_suffix = abs_path.split("training_data/images/")[-1]
                clean_path = f"images/kalapa/{rel_suffix}"
            else:
                clean_path = abs_path
                
            converted_lines.append(f"{clean_path}\t{label}")
            
    with open(output_txt, 'w', encoding='utf-8') as f:
        f.write("\n".join(converted_lines) + "\n")
    print(f"Đã chuẩn hóa {len(converted_lines)} dòng sang: {output_txt}")

raw_train = "/kaggle/working/vietocr_data/train_label.txt"
raw_val = "/kaggle/working/vietocr_data/val_label.txt"

pkg_train = os.path.join(labels_dir, "train_label.txt")
pkg_val = os.path.join(labels_dir, "val_label.txt")

convert_to_relative(raw_train, pkg_train)
convert_to_relative(raw_val, pkg_val)
```
